```python
## How do total monthly sales compare across January, February, and March 2019?
super_market["Date"] = pd.to_datetime(super_market["Date"])
super_market["Month"] = super_market["Date"].dt.month
monthly_sales = super_market[super_market["Month"].isin([1,2,3])].groupby("Month")["Sales"].sum()
monthly_names = {1:"January",2:"February",3:"March"}
monthly_sales = monthly_sales.rename(index=monthly_names)
print(monthly_sales)
```

```python
## Which day of the week generates the highest total sales volume?
super_market["Date"] = pd.to_datetime(super_market["Date"])
super_market["DayOfWeek"] = super_market["Date"].dt.dayofweek
weekly_sales = super_market[super_market["DayOfWeek"].isin([0,1,2,3,4,5,6])].groupby("DayOfWeek")["Sales"].sum()
day_names = {0:"Monday", 1:"Tuesday", 2:"Wednesday", 3:"Thursday", 4:"Friday", 5:"Saturday", 6:"Sunday"}
weekly_sales = weekly_sales.rename(index=day_names)
print(weekly_sales)

weekly_hightest_sale=weekly_sales.idxmax()
print(f"{weekly_hightest_sale}:  weekly hightest sales")
```

```python
## During which hours of the day (morning, afternoon, evening) do peak transactions occur?

# Convert the 'Time' column (which is currently a string like '1:08:00 PM') to datetime objects
# and then extract the hour.
super_market["Peak_Hour"] = pd.to_datetime(super_market["Time"]).dt.hour

# Define a function to categorize the hours into periods
def categorize_hour(hour):
    if 0 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 17:
        return "Afternoon"
    else: # 18 to 23
        return "Evening"

# Apply the categorization to create a new 'Time_Category' column
super_market["Time_Category"] = super_market["Peak_Hour"].apply(categorize_hour)

# Group by the new 'Time_Category' and sum the 'Sales'
sales_by_time_category = super_market.groupby("Time_Category")["Sales"].sum()

# Reorder the categories for better presentation
order = ["Morning", "Afternoon", "Evening"]
sales_by_time_category = sales_by_time_category.reindex(order)

print("Total sales by time of day:")
print(sales_by_time_category)
```

```python
## Do Member customers buy larger quantities on average than Normal customers?
quantity = super_market.groupby("Customer type")["Quantity"].sum()
larger_uantity = quantity.idxmax()
numer_of_quantity = quantity.max()
print(f"larger quantities : {larger_uantity} : {numer_of_quantity}")
print(quantity)
```

```python
## What is the ratio of Member vs. Normal customer transactions per store Branch?

# Group by 'Branch' and 'Customer type' and count the transactions
transactions_by_branch_and_type = super_market.groupby(['Branch', 'Customer type']).size().unstack(fill_value=0)

# Calculate the ratio of Member to Normal transactions per branch
# Replace 0 in 'Normal' column with NaN to avoid division by zero
transactions_by_branch_and_type['Ratio_Member_to_Normal'] = transactions_by_branch_and_type['Member'] / transactions_by_branch_and_type['Normal'].replace(0, float('nan'))

print("Transaction counts and Member vs. Normal ratio per branch:")
print(transactions_by_branch_and_type)
```

```python
## What is the average total spending for transactions occurring before 12:00 PM versus after 5:00 PM?

# Convert 'Time' column to datetime objects and extract the hour
super_market["Peak_Hour"] = pd.to_datetime(super_market["Time"]).dt.hour

# Define a function to categorize the hours based on the problem statement
def Time_Category(hour):
  if hour < 12: # Before 12:00 PM (hours 0-11)
    return "Before_12PM"
  elif hour > 17: # After 5:00 PM (hours 18-23)
    return "After_5PM"
  else: # Hours 12-17 (12:00 PM to 5:00 PM inclusive)
    return "Midday_to_Afternoon"

# Apply the categorization to create a new 'time_category' column
super_market["time_category"] = super_market["Peak_Hour"].apply(Time_Category)

# Calculate the average total spending for each time category
average_spending = super_market.groupby("time_category")["Sales"].mean()

# Print the results for all time categories
print("Average total spending by time category:")
print(average_spending)
```
